# 03 · Explore the codes interactively (Jupyter widgets)

**What this notebook is for.** The browser explorer (`explorer/index.html`) in notebook form. Pick a code, switch views, slide through
stabilizers, and type an error such as `X3 Z7`. The syndrome and, for small codes, the lookup decoder's verdict appear under the picture.

> Widgets need a live kernel. In a saved copy you only see the starting picture. Run the cells to play.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))      # make the repo's qdec package importable
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams["figure.dpi"] = 80
rng = np.random.default_rng(2026)                        # fixed seed: same pictures every run
from qdec.widgets import code_explorer, bp_player, mwpm_player, parse_error, describe_error
from qdec.export import default_replays, lookup_table
from qdec.families import steane
```

## 1. Code explorer
Dropdown = code, buttons = view, slider = which stabilizer to highlight, text box = your error.

```python
code_explorer()
```

## 2. The same thing without widgets (works in a saved notebook too)
`parse_error` reads text like `X3 Z5`; `describe_error` gives the syndrome and the lookup decoder's verdict.

```python
st = steane()
table = lookup_table(st)
for text in ["X3", "Z5", "Y4", "X3 Z5", "X0 X1"]:
    ex, ez = parse_error(text, st.n)
    print(f"{text:6s} ->", describe_error(st, ex, ez, table).replace(chr(10), " | "))
```
